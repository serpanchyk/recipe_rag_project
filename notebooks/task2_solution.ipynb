```python
from openai import AzureOpenAI
import os
from dotenv import load_dotenv
```

```python
endpoint = "https://rag-recipe-resource.openai.azure.com/"
deployment = "gpt-35-turbo"
api_version = "2024-12-01-preview"
load_dotenv()
API_KEY = os.getenv('API_KEY_AZURE')
```

```python
client = AzureOpenAI(
    azure_endpoint=endpoint,
    api_key=API_KEY,
    api_version=api_version
)
```

```python
def response_generator(system_prompt, user_prompt):
    response = client.chat.completions.create(
        model="gpt-35-turbo",
        messages=[
                {"role": "system", "content": system_prompt},
                {"role": "user", "content": user_prompt}
            ],
        response_format={"type": "json_object"}
    )

    return response.choices[0].message.content
```

```python
def cooking_instructions(title):
    system_prompt = """
    You are an expert chef. Your task is to generate detailed cooking instructions
    for the given dish. Your output must be valid JSON according to the schema.
    """

    user_prompt = f"""
    Generate step-by-step cooking instructions for "{title}".
    Respond ONLY in JSON with the following fields:

    {{
        "title": "{title}",
        "recipe": [
            "Step 1: ...",
            "Step 2: ...",
            "Step 3: ..."
        ]
    }}
    """

    return response_generator(system_prompt, user_prompt)
```

```python
def required_ingredients(title):
    system_prompt = """
        You are the assistant chef. Your task is to quickly and accurately list all the necessary ingredients with the quantities for the given dish. The output format must be exclusively JSON, according to the provided schema."""
    user_prompt = f"""
        Respond in JSON format with:
        {{
            "title": "{title}",
            "ingredients": [
                "ingredient1",
                "ingredient2",
                ...
            ]
        }}
        """

    return response_generator(system_prompt, user_prompt)
```

```python
print(cooking_instructions("Holubtsi"))
```

```python
print(required_ingredients("Holubtsi"))
```
